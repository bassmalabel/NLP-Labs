```python
import gensim.downloader as api
from gensim.models import KeyedVectors
import numpy as np
```

```python
# Q1) Using Gensim Library, write the steps to get the vector of the word ‘King’.
#word2vec = api.load("word2vec-google-news-300") this one takes a long time to load
word2vec = api.load("glove-wiki-gigaword-50")
king_vector = word2vec['king']
print(king_vector)
```

```python
# Q2) Get the most similar words to ‘King’.
word2vec.most_similar('king', topn=5)
```

```python
# Q3) Get the similarity between ‘King’ and ‘Queen’.
word2vec.similarity('king', 'queen')
```

```python
# Q4) Perform the word analogy task of 'King' is to 'woman' as 'man' is to??
analogy = word2vec.most_similar(positive=['king', 'woman'], negative=['man'])
print(f"'King' is to 'woman' as 'man' is to: {analogy[0][0]}")
```

```python
# other examples of analogy 
analogy_1 = word2vec.most_similar(positive= ['father', 'woman'], negative= ['man'])
print(f"'Father' is to 'woman' as 'man' is to : {analogy_1[0][0]}")
analogy_2 = word2vec.most_similar(positive=['night', 'sun'], negative= ['moon'])
print(f"'Night' is to 'sun' as 'moon' is to: {analogy_2[0][0]}")
```

### Creating our own word embeddings with Word2Vec

```python
# Q5) Create Your Own Word Embeddings with Word2Vec.
from gensim.utils import simple_preprocess
corpus= [
    'In this chapter we introduce the formal problem of finite Markov decision processes, or finite MDPs, which we try to solve in the rest of the book.',
    'This problem involves evaluative feedback, as in bandits, but also an associative aspect—choosing different actions in different situations.',
    'MDPs are a mathematically idealized form of the reinforcement learning problem for which precise theoretical statements can be made.',
    'We try to convey the wide range of applications that can be formulated as finite MDPs.',
    'As in all of artificial intelligence, there is a tension between breadth of applicability and mathematical tractability.',
    'MDPs are meant to be a straightforward framing of the problem of learning from interaction to achieve a goal.',
]

processed_corpus = [simple_preprocess(sentence) for sentence in corpus]
print(processed_corpus)
```

```python
from gensim.models import Word2Vec

# Train Word2Vec model
word2vec_model = Word2Vec(sentences=processed_corpus, vector_size=50, window=2, min_count=1, workers=-1)

# Save the model
word2vec_model.save("word2vec_small_corpus.model")

# Check the vocabulary
print(f"Vocabulary: {list(word2vec_model.wv.index_to_key)}")
```

```python
problem_w2v = word2vec['problem']
problem_w2v_trained = word2vec_model.wv['problem']

print(f"vector for 'problem' with the pretrained model is: {problem_w2v}, while with the manually trained model is: {problem_w2v_trained}")
```

```python
# finding similar words to problem using both models 

most_sim_pretrained = word2vec.most_similar('problem', topn=3)
most_sim_trained = word2vec_model.wv.most_similar('problem', topn=3)

print(f"top most similar 3 words to problem using the pretrained model are: {most_sim_pretrained}")
print(f"top most similar 3 words to problem using the manually trained model are: {most_sim_trained}")
```

**Comparision between Word2Vec, GloVe and FastText**

* Word2Vec: Great for capturing semantic relationships but doesn’t consider subwords.

* GloVe: Captures global context but ignores subword structure.

* FastText: Combines benefits of Word2Vec with subword information, making it robust for rare and morphologically rich words.

```python
#necessary import 
import spacy
```

Q1) Apply tokenisation and sentence segmentation on the text. 

```python
nlp = spacy.load('en_core_web_sm')
text = 'NLP is amazing. It helps computers understand human language.'

docs = nlp(text)

#tokenization 
print('=== The tokens ===')
for token in docs: 
    print(token.text)
#sentence segmentation 
print('\n=== The sentences ===')
for sent in docs.sents: 
    print(sent)
```

Q2) List all verbs and nouns found in the phrase.

```python
phrase = 'The teacher explains the lesson and the students take notes' 

docs = nlp(phrase)

verbs = [token.text for token in docs if token.pos_ == 'VERB']
nouns = [token.text for token in docs if token.pos_ == 'NOUN']

print('All the verbs of the sentence', verbs)
print('All the nouns of the sentence', nouns)
```

Q3) Identify and classify the named entities in the following text.

```python
text = 'Apple Inc. was founded by Steve Jobs in Cupertino, California on April 1, 1976. The company is now worth over $2 trillion'

docs = nlp(text)

print('=== All the named entities of the sentence are: ===')
for ent in docs.ents: 
    print('the word: ' , ent)
    print('the label: ', ent.label_)
    print('the explanation of the label: ', spacy.explain(ent.label_), '\n')
```

Q4) Remove stop words from the sentence. 

```python
text = 'This is an example showing off stop word filtration.' 

docs = nlp(text) 

filtered_text = [token.text for token in docs if not token.is_stop]

filtered_text
```

Q5) Provide the most similar words to Animal

```python
import gensim.downloader as api 

model = api.load('glove-wiki-gigaword-50') 

most_similar_words = model.most_similar('animal', topn=5)

print(f"The most 5 similar words to 'Animal' are: {most_similar_words}")
```
